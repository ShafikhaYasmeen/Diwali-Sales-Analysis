# Exploratory Data Analysis


```python
df.columns
```

```python
ax=sns.countplot(x= 'Gender', data=df)
for bars in ax.containers:
    ax.bar_label(bars)
```

##### This code calculates total sales amount by gender and then visualizes it using a bar chart.

```python
sales_gen=df.groupby(['Gender'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)
sns.barplot(x='Gender',y='Amount',data=sales_gen)
```

##### From the above graph, it is observed that one female contributes a higher total sales amount compared to the male, indicating a stronger purchasing contribution from that group.

#### AGE

##### how many male and female customers belong to each age group and prints the count on every bar.

```python
ax=sns.countplot(data=df,x='Age Group',hue='Gender')

for bars in ax.containers:
    ax.bar_label(bars)
```

##### This code calculates total sales amount for each age group and then visualizes it using a bar chart.

```python
#Total Amount vs Age Group
sales_age=df.groupby(['Age Group'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)

sns.barplot(x='Age Group',y='Amount', data=sales_age)
```

##### From above graph we can see that most of the buyers are of age group between 26-35yrs female

#### State

```python
sales_state=df.groupby(['State'],as_index=False)['Orders'].sum().sort_values(by='Orders',ascending=False).head(10)

sns.set(rc={'figure.figsize':(15,5)})

sns.barplot(data=sales_state, x='State', y='Orders')
```

##### Average Order Value (AOV) by Gender

```python
aov_gender=df.groupby('Gender', as_index=False).agg(Total_Sales=('Amount','sum'),Total_Orders=('Orders','sum'))

aov_gender['Avg_Order_Value'] = aov_gender['Total_Sales'] / aov_gender['Total_Orders']
```

```python
sns.set(rc={'figure.figsize':(15,10)})
sns.barplot(data=aov_gender,x='Gender',y='Avg_Order_Value')
plt.title("Average Order Value by Gender")
plt.show()
          
```

##### Female customers not only place more orders but also have a higher average order value compared to male customers.

#### Spending Category (Customer Segmentation)

```python
df['Spending_Category'] = pd.cut(df['Amount'],bins=[0, 5000, 15000, 30000, 100000],labels=['Low', 'Medium', 'High', 'Very High'])
ax = sns.countplot(data=df, x='Spending_Category')

for bars in ax.containers:
    ax.bar_label(bars)

```

##### Insight: Most customers fall under the medium spending category, while a smaller group of high-value customers contributes disproportionately to revenue.

#### Occupation vs Average Spending (Not Just Total)

```python
occ_avg = df.groupby('Occupation', as_index=False)['Amount'].mean().sort_values(by='Amount', ascending=False)

sns.barplot(data=occ_avg, x='Occupation', y='Amount')
plt.xticks(rotation=45)
plt.title("Average Spending by Occupation")
plt.show()
```

##### Insight: Professionals in IT, Healthcare, and Aviation sectors have higher average spending per order.

#### Age Group vs Average Spend

```python
age_avg = df.groupby('Age Group', as_index=False)['Amount'].mean().sort_values(by='Amount', ascending=False)

sns.barplot(data=age_avg, x='Age Group', y='Amount')
plt.title("Average Spending by Age Group")
plt.show()
```

##### Insight: Customers aged 26–35 show the highest average spending, making them the most valuable demographic.

#### Repeat Orders Indicator

```python
orders_age = df.groupby('Age Group', as_index=False)['Orders'].sum()

sns.barplot(data=orders_age, x='Age Group', y='Orders')
plt.title("Total Orders by Age Group")
plt.show()
```

#### Insight: The 26–35 age group places the highest number of orders, indicating repeat purchase behavior.

## Key Insights from Diwali Sales Analysis

- Female customers contribute significantly higher revenue and exhibit higher average order value.

- Customers aged 26–35 years are the most valuable segment in terms of both order volume and spending.

- Married women from professional occupations are the primary contributors to festive sales.

- Uttar Pradesh, Maharashtra, and Karnataka lead in total sales and order volume.

- Clothing and Food categories dominate festive purchases, accounting for the highest revenue share.
